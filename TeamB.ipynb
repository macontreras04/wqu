```python
import numpy as np
'''This scrip is based  on the paper "Nine Ways to Implement the Binomial Method
for Option Valuation in MATLAB" by D.J. Highman (2002) SIAM, page 6'''

def binomial_american1(S0, K, T, r, sigma, M,phi,flag_dir):
    dt = T / M
    u = np.exp(sigma * np.sqrt(dt))
    d = 1 / u
    p = (np.exp(r * dt) - d) / (u - d)
    # phi = 1 for call, phi = -1 for put
    # flag_dir = 1 for long, flag_dir = -1 for short

    W=np.zeros((M+1,1))
    W[0] = S0

    for i in range(1, M+1):
        for n in range(i, 0, -1):
            W[n] = u*W[n-1]
        W[0] *= d

    # option call values at time T
    W = np.maximum((W - K)*phi, 0.0)

    # re-trace to get the option value at time 0
    for i in range(M, 0, -1):
        for n in range(i):
            continuation = np.exp(-r * dt) * (p * W[n+1] + (1 - p) * W[n])
            current_price = S0 * u**n * d**(i - 1 - n)
            W[n] = np.maximum(continuation, (current_price - K)*phi)

    price =float(W[0,0])
    final_price = round(price*flag_dir,6)

    return final_price
```

```python
'''The parameters from the question are as follows:'''

S0=100
K=100
T=(3/12)
r=0.05
sigma=0.2
M=100
```

```python
call_ame = binomial_american1(S0, K, T, r, sigma, M,phi=1, flag_dir=1)
print("American Call option price: ", call_ame)

put_ame = binomial_american1(S0, K, T, r, sigma, M,phi=-1, flag_dir=1)
print("American Put option price: ", put_ame)
```
